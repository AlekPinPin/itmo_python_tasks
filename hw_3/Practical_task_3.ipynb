```python
import pandas as pd

## Функция для извлечения имени в столбце Name
def first_name(full_name):
    name_temp=(full_name.split(', ', 1))[1].split()
    name_first_parts = name_temp[1:]
    # Обработка титула при его наличии и выделение имени
    if '.' in name_temp[0]:
        parts_second_parts = name_temp[0].split('.', 1)[1].strip()
        if name_first_parts:
            name = name_first_parts[0]
        else:
            name = parts_second_parts
        name = name.strip('\"\'')
        return name
    else:
        # Обработка, если титула нет
        if name_first_parts:
            name = name_first_parts[0]
            name = name.strip('\"\'')
            return name
    return None

#Считывание из файла train.csv (данные о выживаемости на Титанике)
df_t=pd.read_csv('train.csv')

#Вывод основной информации о датасете: информацию о типах данных, число пропусков, средние значения
print("Содержание датасета:")
print(df_t)

print("\nИнформация о датасете:")
print(df_t.info())

print("\nСредние значения:")
print(df_t.describe())

#Подсчёт процента выживаемости у каждого класса пассажиров (Pclass)
print("\nПроцент выживаемости для каждого класса:")
survival_rate=df_t.groupby('Pclass')['Survived'].mean() * 100
print(survival_rate)

#Вывод самого популярного мужского и самого популярного женского имени на корабле
df_t['FirstName'] = df_t['Name'].apply(first_name)

most_popular_male_name = df_t[df_t['Sex']=='male']['FirstName'].mode()[0]
print(f"\nСамое популярное мужское имя на корабле:{most_popular_male_name}")

most_popular_female_name = df_t[df_t['Sex']=='female']['FirstName'].mode()[0]
print(f"\nСамое популярное женское имя на корабле:{most_popular_female_name}")

#Вывод самого популярного мужского и самого популярного женского имени на корабле в каждом классе
print("\nСамое популярное мужское и самое популярное женское имя в зависимости от класса:")
print(df_t.groupby(['Sex', 'Pclass'])['FirstName'].agg(lambda x: x.mode()[0]))


#Вывод части таблицы с пассажирами, возраст которых больше 44 лет
print("\nПассажиры, кто старше 44 лет:")
print(df_t[df_t['Age']>44])

#Вывод части таблицы с пассажирами, возраст которых меньше 44 лет и которые мужского пола
print("\nПассажиры мужского пола, чей возраст меньше 44 лет:")
print(df_t[
    (df_t['Sex']=='male')&
    (df_t['Age']<44)])

#Вывод количества n-местных кабин (в которых было 2, 3, 4, ... человека)
print("\nКоличество кабин по числу пассажирова:")
cab_size = (df_t['Cabin'].dropna().value_counts()).value_counts().sort_index()
for size, count in cab_size.items():
  if size!=1:
    print(f"Кабин с {size} местами: {count}")
```
