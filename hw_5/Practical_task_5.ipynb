# Данные о выживаемости на Титанике

```python
import pandas as pd
import polars as pl
import numpy as np

!pip install bottleneck
import bottleneck as bn

# Указание источника, где лежит файл train.cvs
url='https://raw.githubusercontent.com/AlekPinPin/itmo_python_tasks/refs/heads/main/train.csv'

#Блок Polars

#Считывание датасет из файла train.csv (это данные о выживаемости на Титанике) с помощью polars
df_tpl=pl.read_csv(url)
print(df_tpl.head())

#Вывод основной информацию о датасете: информацию о типах данных, число пропусков, средние значения и т.д.
#hint: используйте методы и атрибуты датафрейма: describe(), dtypes, null_count(), mean()...
print("\nСредние значения:")
print(df_tpl.describe())

print("\nИнформация о типах:")
print(df_tpl.dtypes)

print("\nКоличество пропусков:")
print(df_tpl.null_count())

print("\nСредние значения:")
print(df_tpl.mean())

#Посчёт количества пассажиров каждого класса (Pclass)
#hint: используйте методы: get_column(), value_counts()...
print("\nКоличество пассажиров каждого класса: ")
print(df_tpl.get_column('Pclass').value_counts())

#Вывод количество выживших мужчин и женщин на корабле
#hint: используйте методы: groupby(), agg(), pl.col(), sum()...
print("\nКоличество выживших мужчин и женщин на корабле:")
print(df_tpl.group_by(['Sex']).agg([pl.col('Survived').sum()]))

#Вывод части таблицы с пассажирами, возраст которых больше 44 лет
#hint: используйте методы: filter(), pl.col()...
print("\nПассажиры,чей возраст больше 44 лет:")
print(df_tpl.filter(pl.col("Age") > 44))


#Ускорение работы с pandas

#Считывание датасета из файла train.csv (это данные о выживаемости на Титанике) с помощью pandas
df_tpd=pd.read_csv(url)
print(df_tpd.head())
#print(bn.bench()) - узнать,какие функции есть


#Посчёт среднего возраста пассажиров и его стандартное отклонение с помощью bottleneck
#hint: используйте методы: bn.nanmean(), bn.nanstd()
print("\nСредний возраст пассажиров:")
print(bn.nanmean(df_tpd['Age']))

print("\nСтандартное отклонение среднего возраста пассажиров:")
print(bn.nanstd(df_tpd['Age']))

#Умножение значения столбца Fare на 1.3 и сохранение результатов как нового столбца Fare_new
#hint: используйте методы: .itertuples() или apply()
df_tpd['Fare_new'] = df_tpd['Fare'].apply(lambda x: x * 1.3)

#Этот вариант значительно быстрее (почти в три раза), чем с использованием apply
#df_tpd['Fare_new'] = np.array(df_tpd['Fare']) * 1.3

#Сторока для проверки добавления столбца
#print(df_tpd.head())
```
