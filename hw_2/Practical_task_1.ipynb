```python
class Chocolate:
    def __init__(self, price: float, addition: str = None):
        if price <= 0:
            raise ValueError("Цена батончика должна быть больше нуля")
        self.price = price
        self.addition = addition

    def show_my_chocolate(self):
        if self.addition:
            print(f"Шоколадный батончик, добавка: {self.addition}")
        else:
            print("Обычный шоколадный батончик")

    def __str__(self):
        return f"Шоколадный батончик, {self.price} рублей"

try:
    chocolate1 = Chocolate(50)
    #chocolate2 = Chocolate(-50)
    chocolate3 = Chocolate(100.2, "орехи с карамелью")
    chocolate4 = Chocolate(30, "воздушный рис")
    #chocolate5 = Chocolate('char')

    for ch in [chocolate1,  chocolate3, chocolate4]:
        print(ch)
        ch.show_my_chocolate()

except ValueError as e:
    print(e)
```
